```python
df['Rating'].unique()
```

```python
df['Rating'] = pd.to_numeric(df['Rating'], errors='coerce')
df['Rating'].dtype
```

```python
df['Reviews'].unique()
```

```python
df[df['Reviews'] =='3.0M']
```

```python
df['Reviews'] = df.Reviews.replace("0.0",0)
df['Reviews'] = df.Reviews.replace("3.0M",3000000.0)
df['Reviews'] = df['Reviews'].astype(float)
df['Reviews'].dtype
```

```python
df['Size'].unique()
```

```python
df['Size'] = df.Size.replace("Varies with device",np.nan)
df['Size'] = df.Size.str.replace("M","000") # All size values became the kilobyte type.
df['Size'] = df.Size.str.replace("k","")
df['Size'] = df.Size.replace("1,000+",1000)
df['Size'] =df['Size'].astype(float)
df['Size'].dtype
```

```python
df['Installs'].unique()
```

```python
df['Installs'] = df.Installs.str.replace(",","")
df['Installs'] = df.Installs.str.replace("+","")
df['Installs'] = df.Installs.replace("Free",np.nan)
df['Installs'] = df['Installs'].astype(float)
df['Installs'].dtype
```

```python
df['Price'].unique()
```

```python
df['Price'] = df.Price.replace("Everyone",np.nan)
df['Price'] = df.Price.str.replace("$","").astype(float)
df['Price'].dtype
```

```python
df.isnull().sum()
```

```python
df['Rating']=df['Rating'].fillna(df['Rating'].median())
```

```python
df['Type'].unique()
```

```python
df.dropna(subset=['Type'],inplace=True)
```

```python
df['Content Rating'].unique()
```

```python
df.dropna(subset=['Content Rating'],inplace=True)
```

```python
sns.histplot(df['Rating'],bins=30,kde=True,color='r')
plt.title('Rating')
plt.grid(True)
plt.show()
```

```python
df.head()
```

```python
sns.countplot(df['Type'])
plt.title('type')
plt.show()
```

```python
plt.figure(figsize=(15,8))

sns.countplot(data=df, y='Category',
              order=df['Category'].value_counts().index,color='r')
plt.title('distribution of app accrose the category',fontsize=16,weight='bold')
plt.show()
```

```python
sns.countplot(df['Content Rating'],color='r')
plt.title('count of content rating',fontsize=10,weight='bold')
plt.show()
```

```python
plt.figure(figsize=(10,12))

sns.barplot(
    data=df,
    y='Category',
    x='Rating',
    order=df.groupby('Category')['Rating'].mean().sort_values(ascending=False).index
)

plt.title('Average Rating by Category')
plt.xlabel('Average Rating')
plt.ylabel('Category')
plt.tight_layout()
plt.show()
```

```python
sns.barplot(x='Type',y='Rating',data=df)
plt.title('Type vs rating',weight='bold')
plt.show()
```

```python
plt.figure(figsize=(10,5))
sns.barplot(x='Content Rating',y='Rating',data=df)
plt.title('content rating vs rating',weight='bold')
plt.show()
```

```python
top10 = df.groupby('Category')['Rating'].mean().nlargest(10).sort_values()

plt.figure(figsize=(10,6))

sns.barplot(x=top10.values, y=top10.index, errorbar=None)

plt.title('Top 10 App Categories by Average Rating')
plt.xlabel('Average Rating')
plt.ylabel('Category')
plt.show()
```
